```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot
```

```python
plt = matplotlib.pyplot
data = pd.read_csv('FuelConsumption.csv')
```

```python
mask = np.random.rand(len(data)) <0.7
train = data[mask]
test = data[~mask]
```

```python
print(f"""train len: {len(train)}
test len: {len(test)}
sum len: {len(test+train)}
train percentage: {round((len(train)/len(data))*100,2)}
test percentage: {round((len(test)/len(data))*100,2)}
""")
```

```python
plt.scatter(train.FUELCONSUMPTION_COMB,train.CO2EMISSIONS,color='black')
plt.scatter(test.FUELCONSUMPTION_COMB,test.CO2EMISSIONS,color='red')
plt.xlabel("FUEL CONSUMPTION")
plt.ylabel("CO2 EMISSIONS")
plt.legend("train vs test")
```
